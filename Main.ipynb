# Q.4 Does the price varies with airlines?

```python
data.groupby('airline')['price'].mean()
```

```python
sns.catplot(x='airline', y='price', hue='class' ,palette='Blues', kind='bar', data=data)
plt.show()
```

# Q.5 Does the ticket price change based on the departure and arrival time?

```python
data.groupby('departure_time')['price'].mean()
```

```python
data.groupby('arrival_time')['price'].mean()
```

```python
sns.catplot(x= 'departure_time', palette='Blues',y='price', kind='bar', data=data)
plt.show()
```

```python
sns.catplot(x= 'arrival_time', palette='Purples',y='price', kind='bar', data=data)
plt.show()
```

```python
sns.relplot(x='arrival_time', y='price', kind='line', data=data, col= 'departure_time')
plt.show()
```
